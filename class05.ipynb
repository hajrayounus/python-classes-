```python
def calculator (a,b,operation):
  if operation == "add":
    return a+b
  elif operation == "sub":
    return a - b
  elif operation == "mul":
    return a*b
  elif operation == "div":
    return a/b

  else:
    return ("invalid function ")

print(calculator(10,20,"add"))
print(calculator(10,20,"sub"))
print(calculator(10,20,"mul"))
print(calculator(10,20,"div"))
```
