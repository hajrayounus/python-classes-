WHY DO WE NEED DICITONARY IN PYTHON :

```python
users = {
    # key     :   value
    "username": "Hajra",
    "age": 20,
    "city": "new york"
}

print(users["username"])
print(users["age"])
print(users["city"])
```

ADDING AND UPDATING PYTHON DICTIONARY

```python
users = {
    # key     :   value
    "username": "Hajra",
    "age": 20,
    "city": "new york"
}

# updating
users["username"] = "uzair"
print(users["username"])

users["age"] = 23
print(users['age'])

users["city"] = "karachi "
print(users["city"])


#adding a key value in our dictionary
users["country"] = "pakistan"
print(users["country"])
```
