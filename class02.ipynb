Chaand Taara Khan is hosting a Kheer (dessert) party at his institute. He needs your help to calculate the total number of people attending the event and the amount of Rabri (in grams and kg) he needs to prepare.

Here’s what you know:

There are 12 faculty members.
There are 4 administrative staff members.
There are 100 students in the institute.
There are 15 people absent.
Per person Rabri is 250g. **bold text**

```python
faculty_member = 12
administrative_staff = 4
students = 100
absence = 15
per_person_rabri = 250
total= faculty_member + administrative_staff + students
total_present_member =total - absence
print (f"total present of member {total_present_member}")
total_rabri = total_present_member * per_person_rabri
print (f"total rabri in gram: {total_rabri}")
print (f"total rabri in kg : {total_rabri // 1000}") # // sa hum point wali value remoe krty han
```
