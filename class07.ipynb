error handling in python


```python
# try:
# try this code
#except:
# do this code
try:
  num = int(input("Enter your number: "))
  number = num / 10
  print (f"result: {number}")

except ZeroDivisionError:
  print("cannt divide by zero. ")
except ValueError:
  print("only numbers are accepted. ")
except Exception as e:
  print (f"an error occured : {e}")
```

ERROR HANDLING IN API REQUEST :

```python
import requests

#API DETAILS
url = "https://chatify.najam.pk/api/v1/sendmessage"
params ={
    "number" : "03278445730",
    "message" : "message from najam ",
    "apikey" : "a2c3e4f5-6789-0123-bcde-456789abcdef"
}

try:
  #send the request
  response = requests.get (url,params=params)
  # check if the request was successful
  if response.status_code == 200:
    print (" message sent successfully : ")
  else :
    print (f"failed to send the messsage . status code : {response.status_code}")
except requests.exceptions.RequestException as e:
  print ("error occured during the request : ")
  print (f"error occured during the request : {e}")
```
