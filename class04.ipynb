list slicing


```python
shopping_item : list[str] = ["cake" , "eggs" , "juice" , "bread", "chips" ]
#slice the chip from list
print (shopping_item[0:4])
print (shopping_item[3])
```

checking membership

```python
shopping_item : list[str] = ["cake" , "eggs" , "juice" , "bread", "chips" ]

print("eggs" in shopping_item)
print ("oil" in shopping_item)
```

adding and removing item from the list


```python
shopping_item : list[str] = ["cake" , "eggs" , "juice" , "bread", "chips" ]
print (shopping_item)

#add flour in list
shopping_item.append("flour")
print (shopping_item)

#remove juice in list
shopping_item.remove("juice")
print (shopping_item)

```

inserting nd removing item

```python
shopping_item : list[str] = ["cake" , "eggs" , "juice" , "bread", "chips" ]
#                           [0       ,  1    ,    2    ,    3   ,   4    ]
#inserting item in list
shopping_item.insert  (2,"flour ")
print("after inserting and before poping ", shopping_item)

#popping item in a list
shopping_item.pop(2)
print("after poping ", shopping_item)
```
