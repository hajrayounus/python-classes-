```python
import random

def generate_math_quiz():
    score = 0
    num_questions = int(input("How many questions would you like to attempt? "))

    for i in range(num_questions):
        num1 = random.randint(1, 10)
        num2 = random.randint(1, 10)
        operator = random.choice(['+', '-', '*'])

        if operator == '+':
            correct_answer = num1 + num2
        elif operator == '-':
            correct_answer = num1 - num2
        else:
            correct_answer = num1 * num2

        question = f"Question {i+1}: What is {num1} {operator} {num2}? "

        try:
            user_answer = int(input(question))
            if user_answer == correct_answer:
                print("Correct!")
                score += 1
            else:
                print(f"Incorrect. The answer was {correct_answer}.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    print(f"\nQuiz finished! You got {score} out of {num_questions} questions correct.")

generate_math_quiz()
```

```python
import random
import time

def roll_dice():
    """Simulates rolling a six-sided die."""
    print("\nRolling the dice...")
    time.sleep(1) # Add a small delay for dramatic effect
    roll = random.randint(1, 6)
    return roll

def play_dice_game():
    print("\n🎲 Welcome to the Super Fun Dice Game! 🎲")
    print("Let's see who can roll the highest or get lucky!")

    num_players = 0
    while num_players < 1:
        try:
            num_players = int(input("How many players are joining? (Enter a number greater than 0): "))
            if num_players < 1:
                print("Please enter a number greater than 0.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    players = [f"Player {i+1}" for i in range(num_players)]
    scores = {player: 0 for player in players}

    game_round = 1
    while True:
        print(f"\n--- Round {game_round} --- ")
        round_rolls = {}

        for player in players:
            input(f"\n{player}, it's your turn! Press Enter to roll the dice...")
            current_roll = roll_dice()
            print(f"\n✨ {player} rolled a {current_roll}! ✨")
            round_rolls[player] = current_roll
            scores[player] += current_roll # Accumulate scores

            if current_roll == 6:
                print("Woohoo! A six! Lucky roll!")
            elif current_roll == 1:
                print("Oops, a one! Better luck next time!")

        print("\n--- Round Scores --- ")
        for player, roll in round_rolls.items():
            print(f"{player}: Rolled {roll} (Total score: {scores[player]}) ")

        highest_roll_this_round = max(round_rolls.values())
        round_winners = [player for player, roll in round_rolls.items() if roll == highest_roll_this_round]

        if len(round_winners) == 1:
            print(f"\n🎉 {round_winners[0]} had the highest roll this round with a {highest_roll_this_round}! 🎉")
        else:
            print(f"\nIt's a tie for the highest roll this round ({highest_roll_this_round}) between {', '.join(round_winners)}!")

        play_again = input("\nPlay another round? (yes/no): ").lower()
        if play_again != 'yes':
            break
        game_round += 1

    print("\n--- Final Scores --- ")
    for player, score in scores.items():
        print(f"{player}: {score}")

    # Determine the overall winner(s)
    if scores:
        max_score = max(scores.values())
        overall_winners = [player for player, score in scores.items() if score == max_score]
        if len(overall_winners) == 1:
            print(f"\n🏆 Congratulations, {overall_winners[0]} is the ultimate winner with a total score of {max_score}! 🏆")
        else:
            print(f"\nIt's a grand tie! The ultimate winners are {', '.join(overall_winners)} with a score of {max_score}! 🥳")
    else:
        print("No scores recorded.")

    print("Thanks for playing! See you next time!\n")

# Start the game
play_dice_game()

```
