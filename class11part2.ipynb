```python
# milestone o1
import random

Num_ROUND = 5
score = 0

print (" welcome to the higher-lower game! ")
print (" _ " * 30)

for round_num in range (1 , Num_ROUND + 1 ):

  computer_number : int = random.randint(1,100)
  user_number : int = random.randint(1,100)

  print ("-" * 50)
  print (f"the user's number is {user_number}")
  print ("-" * 50)

  user_guess = input("do you think your number is higher or lower: ")

  if (
      (user_guess == "higher" and user_number < computer_number) # for higher
      or
      (user_guess == "lower" and user_number > computer_number)  # for lower
      ):
    print (f"you got a point +1 : the computer number is {computer_number} ")
    score += 1
  else :
    print(f"Wrong answer! The computer number was {computer_number}")


print (f"your final score is {score} ")

if score == Num_ROUND:
  print ("wow you played perfectly : ")

elif score == Num_ROUND % 2:
  print ("good job, you played really well: ")

else :
  print ("you can do better next time : ")
```

```python
def access_item(my_list, index):
    if index >= len(my_list):
        return "Index out of range"
    return my_list[index]


def modify_item(my_list, index, new_value):
    if index >= len(my_list):
        return "Index out of range"

    my_list[index] = new_value
    return my_list


def slice_items(my_list, start, end):
    if start >= len(my_list) or end > len(my_list):
        return "Index out of range"

    return my_list[start:end]


def main():

    # Problem 1
    fruit_list = ["apple", "banana", "orange", "grape", "pineapple"]

    print(f"Fruit List: {fruit_list}")

    # Problem 2
    my_list = ["apple", "banana", "orange", "grape", "mango"]

    print(f"\nMy List: {my_list}")

    print("\n1. Access")
    print("2. Modify")
    print("3. Slice")

    choice = input("Choose operation: ")

    # Access
    if choice == "1":

        index = int(input("Enter index: "))

        answer = access_item(my_list, index)

        print("Result:", answer)

    # Modify
    elif choice == "2":

        index = int(input("Enter index: "))
        new_value = input("Enter new value: ")

        answer = modify_item(my_list, index, new_value)

        print("Updated List:", answer)

    # Slice
    elif choice == "3":

        start = int(input("Enter start index: "))
        end = int(input("Enter end index: "))

        answer = slice_items(my_list, start, end)

        print("Sliced List:", answer)

    else:
        print("Invalid choice")


main()
```
