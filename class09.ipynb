Object-Oriented Programming (OOP)
Object-Oriented Programming in Python is a way of writing programs by organizing data and functions into objects.

Real-life example: A car is an object that has properties like color and model and actions like start and stop.

Class
A class is a blueprint or template used to create objects.

Real-life example: A Car class is like a blueprint that describes what every car should have and do.
blueprint like a structure

Object
An object is a real instance created from a class.

Real-life example: A Toyota Corolla can be an object created from the Car class.

```python
# This is a class (blueprint)
class House:
    # Properties
    address: str
    number_of_rooms: int = 4
    number_of_bathrooms: int = 2

    # Constructor function
    def __init__(self, address: str):
        self.address = address
    # method function
    def ring_bell(self):
      print("ding dong")


# This is an object
#                     (aurgemnt )
hajra_house = House("HOUSE no 123")
hajra_house.ring_bell()
print(f"Hajra house address: {hajra_house.address}")

ali_house = House("HOUSE no 321")
print(f"Ali house address: {ali_house.address}")
```

```python
import random

original_number = random.randint(0,99)

while True:
  guessed_number = int(input("ENTER YOUR GUESSED NUMBER: "))

  if (guessed_number < original_number):
    print ("your guess is too low ")
  elif (guessed_number > original_number):
    print ("your guess is too high ")
  else:
    print ("your guess is correct")
    break


```
